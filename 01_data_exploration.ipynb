```python
import pandas as pd
import ast
from collections import Counter

# 1. Load all three datasets (guaranteeing the third is fully included)
df_postings = pd.read_csv('job_postings.csv')
df_clean = pd.read_csv('final_clean_jobs_dataset.csv')
df_ai = pd.read_csv('AI_Job_Market_Trends_2026.csv')

print("Loaded all three successfully:")
print(f"- Job Postings Shape: {df_postings.shape}")
print(f"- Clean Jobs Shape: {df_clean.shape}")
print(f"- AI Market Trends Shape: {df_ai.shape}\n")

# 2. Extract skill distribution from AI Job Market Trends
skill_cols = [col for col in df_ai.columns if col.startswith('skills_')]
ai_skill_counts = df_ai[skill_cols].sum().reset_index()
ai_skill_counts.columns = ['skill', 'count']
ai_skill_counts['skill'] = ai_skill_counts['skill'].str.replace('skills_', '').str.capitalize()
ai_skill_counts = ai_skill_counts.sort_values(by='count', ascending=False)

print("--- AI Job Market Trends Skill Distribution ---")
print(ai_skill_counts)

# 3. Extract skills/languages from Clean Jobs Dataset
all_clean_skills = []
for skills_str in df_clean['skills'].dropna():
    try:
        parsed = ast.literal_eval(skills_str)
        if isinstance(parsed, list):
            all_clean_skills.extend([s.lower().strip() for s in parsed])
    except:
        pass

clean_skill_freq = Counter(all_clean_skills).most_common(10)
print("\n--- Clean Jobs Dataset Skill/Language Counts ---")
for skill, count in clean_skill_freq:
    print(f"{skill}: {count}")

# 4. Check columns of Job Postings to see what skill/language data it contributes
print("\n--- Job Postings Columns ---")
print(list(df_postings.columns))
```
